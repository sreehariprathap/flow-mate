### Project Setup and Dependencies : Prisma

1. Install Prisma CLI
```bash
npm i --save-dev prisma
npm i @prisma/client
```

2. Initialize Prisma with postgresql
```bash
npx prisma init --datasource-provider postgresql
```

3. Create the models in `schema.prism'

4. Run the migration
```bash
npx prisma migrate dev --name init
```
5. To run the studio
```bash
npx prisma studio
```
