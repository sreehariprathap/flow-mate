### Project Setup and Dependencies : Next JS, Shadcn, Lucide React, Clerk

To set up the project and install the necessary dependencies, follow these steps:

1. **Create a new Next.js application**:
    We used the following command to create a new Next.js application with version 14.2.5.
    ```bash
    npx create-next-app@14.2.5
    ```

2. **Add Shadcn**:
    We added Shadcn to our project using the following command:
    ```bash
    npx shadcn@2.1.0 add
    ```

3. **Install Lucide React**:
    We installed Lucide React, a library for icons, using the following command:
    ```bash
    npm i lucide-react
    ```

4. **Authentication**:
    To handle authentication, we installed the Clerk package for Next.js:
    ```bash
    npm install @clerk/nextjs
    ```
